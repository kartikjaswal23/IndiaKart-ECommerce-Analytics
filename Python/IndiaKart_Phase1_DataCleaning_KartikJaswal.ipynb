## Step 13: Referential Integrity Validation

Referential integrity checks were performed to confirm that foreign-key values in transactional tables have matching records in their parent tables.

The following relationships were validated:

- Order Items → Orders
- Payments → Orders
- Returns → Orders
- Orders → Customers
- Order Items → Products

Records without a corresponding parent record would be treated as orphan records and flagged as data-quality issues.

```python
integrity_checks = {
    "Order Items → Orders": (
        order_items['order_id'],
        orders['order_id']
    ),
    
    "Payments → Orders": (
        payments['order_id'],
        orders['order_id']
    ),
    
    "Returns → Orders": (
        returns['order_id'],
        orders['order_id']
    ),
    
    "Orders → Customers": (
        orders['customer_id'],
        customers['customer_id']
    ),
    
    "Order Items → Products": (
        order_items['product_id'],
        products['product_id']
    )
}

integrity_results = []

for relationship, (child, parent) in integrity_checks.items():
    
    orphan_count = (~child.isin(parent)).sum()
    
    integrity_results.append({
        "Relationship": relationship,
        "Records_Checked": len(child),
        "Orphan_Records": orphan_count,
        "Integrity_Status": "PASS" if orphan_count == 0 else "FAIL"
    })

integrity_results = pd.DataFrame(integrity_results)

display(integrity_results)
```

### Referential Integrity Findings

All tested relationships passed the referential integrity checks.

- All 100,000 order-item records have matching orders.
- All 50,000 payment records have matching orders.
- All 10,000 return records have matching orders.
- All 50,000 orders are linked to valid customers.
- All 100,000 order-item records reference valid products.
- No orphan records were identified.

This confirms that the key relationships required for cross-table business analysis are structurally consistent.

## Step 17: Customer Order Count Validation

The `customers` table contains a pre-calculated `total_orders` field. To validate this metric, actual order counts were independently calculated from the orders table and compared with the customer-level values.

This check ensures that customer lifetime statistics are consistent with the underlying transaction data.

```python
actual_order_counts = (
    orders.groupby('customer_id')
    .size()
    .reset_index(name='actual_order_count')
)

customer_validation = customers.merge(
    actual_order_counts,
    on='customer_id',
    how='left'
)

customer_validation['actual_order_count'] = (
    customer_validation['actual_order_count']
    .fillna(0)
    .astype(int)
)

customer_validation['order_count_difference'] = (
    customer_validation['total_orders']
    - customer_validation['actual_order_count']
)

mismatched_customers = customer_validation[
    customer_validation['order_count_difference'] != 0
]

print("Total Customers:", len(customer_validation))
print("Matching Order Counts:",
      (customer_validation['order_count_difference'] == 0).sum())

print("Mismatched Order Counts:",
      len(mismatched_customers))
```

# Phase 1 — Data Quality Report

## Executive Summary

The IndiaKart dataset contains 8 interconnected tables covering customers, orders, order items, products, payments, returns, inventory, and suppliers. A total of 222,200 records were reviewed during the data quality assessment.

Overall, the dataset is structurally strong. No duplicate rows, duplicate primary identifiers, orphan records, invalid order dates, or invalid numeric values were identified. All relevant date fields were standardized to datetime format, and key relationships between customers, orders, products, payments, and returns were successfully validated.

However, several cross-table inconsistencies were identified in the returns data and should be considered when interpreting return-related business metrics.

## Key Data Quality Findings

### 1. Missing Values
Six of the eight datasets contain no missing values.

The `orders.delivered_date` field contains 17,501 missing values (35%). These values were retained because they occur among non-delivered order statuses and can represent valid business conditions.

The `payments.refund_date` field is entirely blank. Investigation showed that `refund_amount` is ₹0 across all 50,000 payment records, indicating that no refund transactions are recorded in the payments table.

### 2. Duplicate and Referential Integrity Checks
No completely duplicated records were identified across any of the eight datasets.

All tested primary identifiers are unique, including customer, order, item, product, payment, return, and supplier IDs.

All tested relationships passed referential integrity validation. No orphan records were found between order items and orders, payments and orders, returns and orders, orders and customers, or order items and products.

### 3. Order Value Review
A total of 312 orders have a final value above ₹5,00,000 and were flagged for review.

Using the IQR method, 5,243 orders (10.49%) were identified as statistical outliers. These transactions were retained because high-value orders may represent legitimate purchases rather than data errors.

### 4. Returns Data Quality Issues
Returns data contains the most significant consistency issues identified during Phase 1.

- 1,006 return records have a return date earlier than the recorded delivery date.
- 5,673 return records are linked to orders without a recorded delivery date.
- 6,067 orders appearing in the returns table are not marked as `Returned` in the orders table.
- All 3,933 orders marked as `Returned` in the orders table do have corresponding records in the returns table.

These records were flagged rather than modified because the correct source-of-truth business rules should be confirmed before altering transactional data.

## Business Risk Indicators

Two early business risks were identified during the quality assessment:

- Payment Failure Rate: 3.50%, above the management target of less than 2%.
- Cancellation Rate: 11.99%, above the management target of less than 10%.

These are valid business events rather than data-cleaning errors and should be investigated further during exploratory and KPI analysis.

Inventory currently contains 2 Out of Stock products and 76 Low Stock products. In addition, 457 customers (4.57%) have placed zero orders, which is considered valid because registered customers may not yet have completed a purchase.

## Cleaning Decisions

No rows were deleted during Phase 1.

Date fields were standardized, high-value transactions were flagged rather than removed, logically valid missing values were retained, and identified returns inconsistencies were documented for further investigation.

The cleaned datasets are suitable for exploratory analysis, with caution required when interpreting return-related metrics because of the identified cross-table inconsistencies.
